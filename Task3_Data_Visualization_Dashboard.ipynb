```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("cleaned_data.csv", parse_dates=["Order Date", "Ship Date"])
df["Month"] = df["Order Date"].dt.month
df["Year"] = df["Order Date"].dt.year
sns.set_style("whitegrid")
```

```python
cat = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
plt.figure(figsize=(7,5))
plt.bar(cat.index, cat.values, color=["#1f77b4", "#ff7f0e", "#2ca02c"])
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

```python
monthly = df.groupby("Month")["Sales"].sum()
plt.figure(figsize=(8,5))
plt.plot(monthly.index, monthly.values, marker="o", color="teal", label="Sales")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(range(1, 13))
plt.legend()
plt.show()
```

```python
seg = df["Segment"].value_counts()
plt.figure(figsize=(6,6))
plt.pie(seg.values, labels=seg.index, autopct="%1.1f%%", colors=["#1f77b4", "#ff7f0e", "#2ca02c"])
plt.title("Orders by Customer Segment")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df[df["Sales"] < 1000]["Sales"], bins=40, color="purple", label="Sales")
plt.title("Sales Distribution (orders under 1000)")
plt.xlabel("Sales")
plt.ylabel("Number of Orders")
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.scatterplot(data=df, x="Order Date", y="Sales", hue="Region", alpha=0.6)
plt.title("Sales Over Time by Region")
plt.xlabel("Order Date")
plt.ylabel("Sales")
plt.legend(title="Region")
plt.show()
```

Insights
- Technology leads sales by category.

- Consumer is the largest customer segment by orders.
