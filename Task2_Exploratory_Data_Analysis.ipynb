```python
import pandas as pd
df = pd.read_csv("cleaned_data.csv", parse_dates=["Order Date", "Ship Date"])
df.describe()
```

```python
# Outliers in Sales (IQR method)
q1, q3 = df["Sales"].quantile([0.25, 0.75])
iqr = q3 - q1
outliers = df[(df["Sales"] < q1 - 1.5*iqr) | (df["Sales"] > q3 + 1.5*iqr)]
print("Outliers:", len(outliers))
print(outliers.sort_values("Sales", ascending=False)[["Product Name", "Category", "Sales"]].head(5))
```

```python
df["Month"] = df["Order Date"].dt.month_name()
df["Year"] = df["Order Date"].dt.year

print("Sales by Category:\n", df.groupby("Category")["Sales"].sum().sort_values(ascending=False), "\n")
print("Sales by Region:\n", df.groupby("Region")["Sales"].sum().sort_values(ascending=False), "\n")
print("Sales by Segment:\n", df.groupby("Segment")["Sales"].sum().sort_values(ascending=False), "\n")
print("Top 5 Sub-Categories:\n", df.groupby("Sub-Category")["Sales"].sum().sort_values(ascending=False).head(5), "\n")
print("Sales by Year:\n", df.groupby("Year")["Sales"].sum(), "\n")
print("Best Months:\n", df.groupby("Month")["Sales"].sum().sort_values(ascending=False).head(3))
```

Task 2: Exploratory Data Analysis

Key Findings
- Technology is the top category by sales (about 827K), followed by Furniture and Office Supplies.
- West is the strongest region (about 710K); South is the weakest (about 389K).
- Consumer is the most valuable segment (about 1.15M), ahead of Corporate and Home Office.
- Phones, Chairs and Storage are the top-selling sub-categories.
- Sales dipped slightly in 2016, then grew strongly in 2017 and 2018.
- November, December and September are the best months, showing a year-end seasonal peak.
- 1,145 outliers were found using the IQR method. They are high-value Technology orders (e.g., Cisco TelePresence, Canon copiers), so they are genuine sales and were kept.
- Sales has almost no correlation with Postal Code (-0.03).

Conclusion: Sales are driven by Technology and the Consumer segment, peak at year-end, and the South region has the most room for growth.
